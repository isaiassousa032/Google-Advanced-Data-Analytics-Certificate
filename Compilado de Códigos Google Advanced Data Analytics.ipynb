# Annotated follow-along guide: EDA using basic data functions with Python

```python
import pandas as pd
import numpy as np
import datetime as dt
import matplotlib.pyplot as plt
```

```python
# Leia o conjunto de dados de raios de 2018. # Read in the 2018 lightning strike dataset.
df = pd.read_csv('eda_using_basic_data_functions_in_python_dataset1.csv')
```

```python
# Inspecione as 10 primeiras linhas. # Inspect the first 10 rows.
df.head(10)
```

```python
df.shape
```

```python
# Get more information about the data, including data types of each column
df.info()
```

```python
# Convert date column to datetime
df['date']= pd.to_datetime(df['date'])
```

```python
# Calculate days with most lightning strikes.
# Calcular os dias com maior número de raios.
df.groupby(['date']).sum().sort_values('number_of_strikes', ascending=False).head(10) 
```

```python
# Criar uma nova coluna `mês` # Create a new `month` column 
df['month'] = df['date'].dt.month
df.head()
```

```python
# Calcular o número total de raios por mês # Calculate total number of strikes per month
df.groupby(['month']).sum().sort_values('number_of_strikes', ascending=False).head(12)
```

```python
# Crie uma nova coluna `month_txt` # Create a new `month_txt` column.
df['month_txt'] = df['date'].dt.month_name().str.slice(stop=3)
df.head()
```

```python
# Crie um novo dataframe auxiliar para plotagem. # Create a new helper dataframe for plotting.
df_by_month = df.groupby(['month','month_txt']).sum().sort_values('month', ascending=True).head(12).reset_index()
df_by_month
```

```python
# Criar um gráfico de barras
plt.bar(x=df_by_month['month_txt'],height= df_by_month['number_of_strikes'], label="Number of strikes")
plt.plot()

plt.xlabel("Months(2018)")
plt.ylabel("Number of lightning strikes")
plt.title("Number of lightning strikes in 2018 by months")
plt.legend()
plt.show()
```

```python
# Step 1: Use pd.to_datetime() to convert Date Joined column to datetime 
# Step 2: Update the column with the converted values

companies['Date Joined'] = pd.to_datetime(companies['Date Joined'])
```

```python
# Use .info() to confirm that the update actually took place

companies.info()
```

```python
#Step 1: Use .dt.year to extract year component from Date Joined column # Etapa 1: Use .dt.year para extrair o componente do ano da coluna Data de Junção
#Step 2: Add the result as a new column named Year Joined to the DataFrame # Step 2: Add the result as a new column named Year Joined to the DataFrame

companies['Year Joined'] = companies['Date Joined'].dt.year
```

```python
# Use .head() to confirm that the new column did get added

companies.head()
```

```python
# Sample the data

companies_sampled = companies.sample(n=50, random_state= 42)
```

```python
# Prepare data for plotting

# Criar nova coluna `years_till_unicorn`
companies_sampled['years_till_unicorn'] = companies_sampled['Year Joined'] - companies_sampled['Year Founded']

# Agrupe os dados por `Industry`. Para cada indústria, obtenha o valor máximo na coluna `years_till_unicorn`
grouped = (companies_sampled[["Industry", "years_till_unicorn"]]
           .groupby("Industry")
           .max()
           .sort_values(by="years_till_unicorn")
          )
grouped
```

```python
# Criar gráfico de barras 
# com os diversos setores como categorias das barras 
# e o tempo necessário para atingir o status de unicórnio como a altura das barras


# Create bar plot
# with the various industries as the categories of the bars
# and the time it took to reach unicorn status as the height of the bars
plt.bar(grouped.index, grouped['years_till_unicorn'])

# Definir Título
# Set title
plt.title("Bar plot of maximum years taken by company to become unicorn per industry (from sample)")

# Definir o rótulo do eixo x
# Set x-axis label
plt.xlabel("Industry")

# Definir o rótulo do eixo y
plt.ylabel("Maximum number of years")

# Rotacionar os rótulos no eixo x para evitar sobreposição nas posições do texto.
# Rotate labels on the x-axis as a way to avoid overlap in the positions of the text
plt.xticks(rotation=45, horizontalalignment='right')

# Exibir o gráfico
# Display the plot
plt.show()
```

```python
# Criar uma coluna representando a avaliação da empresa como dados numéricos
# Create a column representing company valuation as numeric data

companies_sampled['valuation_billions'] = companies_sampled['Valuation']

companies_sampled['valuation_billions'] = companies_sampled['valuation_billions'].str.replace('$', '')

companies_sampled['valuation_billions'] = companies_sampled['valuation_billions'].str.replace('B', '')

companies_sampled['valuation_billions'] = companies_sampled['valuation_billions'].astype('int')
companies_sampled.head()
```

```python
# Prepare data for modeling
grouped = (companies_sampled[["Industry", "valuation_billions"]]
           .groupby("Industry")
           .max()
           .sort_values(by="valuation_billions")
          )
grouped
```

```python
# Criar gráfico de barras com os diferentes setores como categorias das barras e a avaliação máxima de cada setor como altura das barras
# Create bar plot with the various industries as the categories of the bars and the maximum valuation for each industry as the height of the bars
plt.bar(grouped.index, grouped["valuation_billions"])

# Definir título
# Set title
plt.title("Bar plot of maximum unicorn company valuation per industry (from sample)")

# Definir rótulo do eixo x
# Set x-axis label
plt.xlabel("Industry")

# Definir rótulo do eixo y
# Set y-axis label
plt.ylabel("Maximum valuation in billions of dollars")

# Rotacionar rótulos no eixo x para evitar sobreposição de texto
# Rotate labels on the x-axis as a way to avoid overlap in the positions of the text  
plt.xticks(rotation=45, horizontalalignment='right')#

# Exibir o gráfico
plt.show()
```
