Use the `to_datetime()` function from the `pandas` library  to convert the `Date Joined` column to datetime. This splits each value into year, month, and date components. This is an important step in data cleaning, as it makes the data in this column easier to use in tasks you may encounter. ### Visualize the time it took companies to reach unicorn status
### Visualize o tempo que as empresas levaram para atingir o status de unicórnio.

Visualize o tempo máximo que as empresas levaram para atingir o status de unicórnio em cada setor representado na amostra. Para criar um gráfico de barras para visualizar isso, use a função `bar()` do pacote. Módulo `matplotlib.pyplot`. Primeiro, você precisará preparar os dados.

Visualize the longest time it took companies to reach unicorn status for each industry represented in the sample. To create a bar plot to visualize this, use the `bar()` function from the `matplotlib.pyplot` module. You'll first need to prepare the data.

```python
# Prepare data for plotting

# Criar nova coluna `years_till_unicorn`
companies_sampled['years_till_unicorn'] = companies_sampled['Year Joined'] - companies_sampled['Year Founded']

# Agrupe os dados por `Industry`. Para cada indústria, obtenha o valor máximo na coluna `years_till_unicorn`
grouped = (companies_sampled[["Industry", "years_till_unicorn"]]
           .groupby("Industry")
           .max()
           .sort_values(by="years_till_unicorn")
          )
grouped
```

```python
# Criar gráfico de barras 
# com os diversos setores como categorias das barras 
# e o tempo necessário para atingir o status de unicórnio como a altura das barras


# Create bar plot
# with the various industries as the categories of the bars
# and the time it took to reach unicorn status as the height of the bars
plt.bar(grouped.index, grouped['years_till_unicorn'])

# Definir Título
# Set title
plt.title("Bar plot of maximum years taken by company to become unicorn per industry (from sample)")

# Definir o rótulo do eixo x
# Set x-axis label
plt.xlabel("Industry")

# Definir o rótulo do eixo y
plt.ylabel("Maximum number of years")

# Rotacionar os rótulos no eixo x para evitar sobreposição nas posições do texto.
# Rotate labels on the x-axis as a way to avoid overlap in the positions of the text
plt.xticks(rotation=45, horizontalalignment='right')

# Exibir o gráfico
# Display the plot
plt.show()
```

### Visualize the maximum unicorn company valuation per industry

Visualize unicorn companies' maximum valuation for each industry represented in the sample. To create a bar plot to visualize this, use the `bar()` function from the `matplotlib.pyplot` module. Before plotting, create a new column that represents the companies' valuations as numbers (instead of strings, as they're currently represented). Then, use this new column to plot your data.

```python
# Criar uma coluna representando a avaliação da empresa como dados numéricos
# Create a column representing company valuation as numeric data

companies_sampled['valuation_billions'] = companies_sampled['Valuation']

companies_sampled['valuation_billions'] = companies_sampled['valuation_billions'].str.replace('$', '')

companies_sampled['valuation_billions'] = companies_sampled['valuation_billions'].str.replace('B', '')

companies_sampled['valuation_billions'] = companies_sampled['valuation_billions'].astype('int')
companies_sampled.head()
```

```python
# Prepare data for modeling
grouped = (companies_sampled[["Industry", "valuation_billions"]]
           .groupby("Industry")
           .max()
           .sort_values(by="valuation_billions")
          )
grouped
```

```python
# Criar gráfico de barras com os diferentes setores como categorias das barras e a avaliação máxima de cada setor como altura das barras
# Create bar plot with the various industries as the categories of the bars and the maximum valuation for each industry as the height of the bars
plt.bar(grouped.index, grouped["valuation_billions"])

# Definir título
# Set title
plt.title("Bar plot of maximum unicorn company valuation per industry (from sample)")

# Definir rótulo do eixo x
# Set x-axis label
plt.xlabel("Industry")

# Definir rótulo do eixo y
# Set y-axis label
plt.ylabel("Maximum valuation in billions of dollars")

# Rotacionar rótulos no eixo x para evitar sobreposição de texto
# Rotate labels on the x-axis as a way to avoid overlap in the positions of the text  
plt.xticks(rotation=45, horizontalalignment='right')#

# Exibir o gráfico
plt.show()
```
